```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

titanic = sns.load_dataset('titanic')

plt.figure(figsize=(12, 8))

ax = sns.boxplot(data=titanic, x='sex', y='age', hue='survived', palette='Set2')

plt.title('Distribution of Age by Gender and Survival Status', fontsize=15)
plt.xlabel('Gender (Sex)', fontsize=12)
plt.ylabel('Age', fontsize=12)

handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, ['No (0)', 'Yes (1)'], title='Survived')

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=titanic, x='age', hue='survived', kde=True, palette='viridis')
plt.title('Age Distribution by Survival Status', fontsize=15)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend(title='Survived', labels=['No (0)', 'Yes (1)'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=titanic, x='sex', hue='survived', palette='coolwarm')
plt.title('Survival Count by Gender', fontsize=15)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.legend(title='Survived', labels=['No (0)', 'Yes (1)'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Observations & Inferences (DSBDA Practical A-9)

Based on the box plot, histogram, and count plot, here are the key statistical inferences and observations for the assignment write-up:

#### Observations from Box Plot (Distribution of Age by Gender and Survival Status):

*   **Age Distribution:**
    *   The median age for both males and females across survival groups is centered around **30–35 years**.
    *   The **Interquartile Range (IQR)**, representing the middle 50% of the data, spans roughly from **20 to 40 years**, indicating that the majority of passengers were young to middle-aged adults.

*   **Survival vs. Non-Survival:**
    *   **Males:** The median age of non-surviving males is slightly higher than that of survivors, suggesting older men were less likely to survive.
    *   **Females:** Survivors generally belong to a slightly younger age group compared to non-survivors.
    *   **Overall Trend:** Females across all age groups show a significantly higher survival rate compared to males, consistent with the 'women and children first' policy.

*   **Outliers & Children:**
    *   **Outliers:** There is a notable presence of outliers in the higher age brackets (70–80 years), particularly among males who did not survive.
    *   **Children:** The lower whiskers of the box plots extend close to 0, confirming the presence of infants and children in both surviving and non-surviving groups.

#### Observations from Age Distribution Histogram:

*   **Younger Passengers:** A higher density of younger individuals (especially children) appear in the 'Survived' category.
*   **Elderly Passengers:** Very few passengers in older age groups (70+) survived.
*   **General Age Groups:** The non-survived group tends to have a broader distribution across middle ages, while the survived group shows peaks at younger ages.

#### Observations from Survival Count by Gender:

*   **Male Non-Survivors:** A significantly larger number of males did not survive compared to females.
*   **Female Survivors:** The number of female survivors is notably higher than male survivors, despite there being fewer total female passengers.
*   **Confirmation of 'Women and Children First':** This plot clearly reinforces the observation that females had a much higher survival rate than males.

#### Key Takeaway (Overall):

*   The visualizations collectively highlight that **younger individuals and females** had higher survival probabilities. Conversely, **older males** represent the demographic with the lowest survival rate in this dataset.
