```python
import pandas as pd

df=pd.read_csv('/content/academic_performance_students_raw.csv')
display(df.head(10))
```

```python
df.info()
```

```python
#To get the descriptive statistics for all numerical columns
df.describe()
```

```python
#Unique Values of  Gender and its counts
df['Gender'].value_counts()
```

```python
#Unique Values of Final_Result and its Counts
df['Final_Result'].value_counts()
```

```python
#Unique Values of Name and its counts
df['Name'].value_counts()
```

```python
#Correcting invalid 'Attendance_%' values
#Set negative attendance to 0
df.loc[df['Attendance_%']<0,'Attendance_%']=0

#Set Attendance greater than 100 to be 100
df.loc[df['Attendance_%']>100,'Attendance_%']=100

#Verification
display(df['Attendance_%'].describe())
```

```python
#Imputation of Age,Math,Science and English
#Impute missing Age and the subject  values with the mean

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Math'] = df['Math'].fillna(df['Math'].mean())
df['Science'] = df['Science'].fillna(df['Science'].mean())
df['English'] = df['English'].fillna(df['English'].mean())

df.info()
```

```python
df.describe()
```

```python
#Convert Gender and Final Result to category data type
df['Gender']=df['Gender'].astype('category')
df['Final_Result']=df['Final_Result'].astype('category')
df.info()
```

```python
#Apply one hot encoding to Gender and Final Result
#drop_first=True is used to avoid multicollinearity for binary variables
df=pd.get_dummies(df,columns=['Gender','Final_Result'],drop_first=True)
display(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

#Initialize the MinMaxScaler
scaler=MinMaxScaler()

#List of numerical columns to scale
numerical_cols=['Age','Attendance_%','Math','Science','English']

#Apply the MinMax scaling to the selected numerical columns
df[numerical_cols]=scaler.fit_transform(df[numerical_cols])

df.head()
```

```python
df.describe()
```

```python
columns_to_check = ['Age', 'Math', 'Science', 'English']
```

```python
for column in columns_to_check:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap the outliers
    df.loc[df[column] < lower_bound, column] = lower_bound
    df.loc[df[column] > upper_bound, column] = upper_bound

df.describe()
```
